### **Question 15**
'Runtime' column se 'min' text ko hata kar use proper integer (numeric) data type mein convert karein.

### **Question 16**
'Gross' column mein jo missing values (NaN) hain, unko pooray column ke median value se fill (impute) karein.

```python
gross_num = pd.to_numeric(df["Gross"].astype(str).str.replace(",", "", regex=False), errors="coerce")
gross_median = gross_num.median()
df["Gross"] = gross_num.fillna(gross_median)
df["Gross"].isna().sum()
```

### **Question 32**

### **Question 33**
Aise actors (Star1) ki list nikalein jinhone 'Action' ya 'Sci-Fi' genre mein kaam kiya hai AUR unki movies ki 'IMDB_Rating' kam az kam 8.0 hai.

```python
genre = df["Genre"].fillna("")
rating = pd.to_numeric(df["IMDB_Rating"], errors="coerce")
mask = genre.str.contains(r"Action|Sci-Fi", case=False, regex=True) & (rating >= 8.0)
df.loc[mask, ["Star1", "Series_Title", "Genre", "IMDB_Rating"]].drop_duplicates().sort_values("Star1")
```

```python
year_num = pd.to_numeric(df["Released_Year"], errors="coerce")
mask = year_num.between(2010, 2020) & df["Certificate"].isin(["UA", "A"])
df.loc[mask]
```

### **Question 35**
Dataset mein 'Meta_score' 90 se zyada aur 'IMDB_Rating' 8.5 se kam wali unique movies kaunsi hain?

```python
meta = pd.to_numeric(df["Meta_score"], errors="coerce")
rating = pd.to_numeric(df["IMDB_Rating"], errors="coerce")
df.loc[(meta > 90) & (rating < 8.5), ["Series_Title", "Meta_score", "IMDB_Rating"]].drop_duplicates()
```

### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
plot_df = df.copy()
plot_df["Gross_num"] = pd.to_numeric(plot_df["Gross"].astype(str).str.replace(",", "", regex=False), errors="coerce")
plot_df["Genre_group"] = np.select(
    [plot_df["Genre"].fillna("").str.contains("Drama", case=False),
     plot_df["Genre"].fillna("").str.contains("Action", case=False)],
    ["Drama", "Action"], default="Other"
)
sns.boxplot(data=plot_df[plot_df["Genre_group"].isin(["Drama", "Action"])], x="Genre_group", y="Gross_num")
plt.title("Gross Revenue: Drama vs Action")
plt.xlabel("Genre")
plt.ylabel("Gross Revenue")
plt.tight_layout()
plt.show()
```

### **Question 37**
Har Director ki un movies ka count maloom karein jinhone 'No_of_Votes' 500,000 (5 lakh) se zyada hasil kiye hain.

```python
votes = pd.to_numeric(df["No_of_Votes"], errors="coerce")
high_vote_director_counts = df.loc[votes > 500_000].groupby("Director")["Series_Title"].count().sort_values(ascending=False)
high_vote_director_counts
```

### **Question 38**
Un movies ke naam aur ratings batayein jinka 'Released_Year' 2000 se pehle ka hai AUR unka 'Gross' unke overall median gross se zyada hai.

```python
year_num = pd.to_numeric(df["Released_Year"], errors="coerce")
gross_num = pd.to_numeric(df["Gross"].astype(str).str.replace(",", "", regex=False), errors="coerce")
mask = (year_num < 2000) & (gross_num > gross_num.median())
df.loc[mask, ["Series_Title", "Released_Year", "IMDB_Rating", "Gross"]].sort_values("Gross", ascending=False)
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
top5 = df.assign(Rating_num=pd.to_numeric(df["IMDB_Rating"], errors="coerce"),
                 Meta_num=pd.to_numeric(df["Meta_score"], errors="coerce")).nlargest(5, "Rating_num").copy()
top5["Rating_scaled_100"] = top5["Rating_num"] * 10
compare = top5.set_index("Series_Title")[["Meta_num", "Rating_scaled_100"]]
compare.plot(kind="bar", figsize=(11, 6))
plt.title("Top 5 Highest-Rated Movies: Meta Score vs IMDb Rating (Scaled to 100)")
plt.xlabel("Movie")
plt.ylabel("Score out of 100")
plt.xticks(rotation=35, ha="right")
plt.legend(["Meta Score", "IMDb Rating × 10"])
plt.tight_layout()
plt.show()
```
