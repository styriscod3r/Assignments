### Q5

```python
total = 0

with open("numbers.txt","r") as file:
    for line in file:
        line =line.strip()
        if line:
            total +=int(line)

print("Sum of all numbers =", total)
```

### Q7 (Medium)
`with open()` statement ka istemal karte hue ek file `data.txt` read karein aur explain karein ki `with` ka kya faida hai.

```python
# Making data.txt first so the code runs without error
with open("data.txt", "w") as f:
    f.write("This is data.txt file.\nPython file handling is easy!")

# Reading it using "with"
with open("data.txt","r") as file:
    data = file.read()
    print(data)

# Why do we use "with"?
# 1. It closes the file automatically, so we don't need file.close().
# 2. The file is closed even if an error happens in the middle.
# 3. Code looks shorter and cleaner.
# 4. It avoids problems like memory leaks or locked files
#    that happen when we forget to close a file.
```

### Q9
Ek file `source.txt` ka data read karke `destination.txt` file mein copy karein.

```python
# Making source.txt first so the code runs
with open("source.txt","w") as f:
    f.write("This text will be copied.\nSecond line of source file.")

# Read from source
with open("source.txt","r") as src:
    data = src.read()

# Write into destination
with open("destination.txt","w") as dest:
    dest.write(data)

# Check if the copy worked
with open("destination.txt","r") as dest:
    print("Content of destination.txt:")
    print(dest.read())
```

### Q14
Ek program banayein jisme `finally` block ho, jo humesha execute ho chahe error aaye ya na aaye.

```python
# Case 1: no error
try:
    print("Case 1: dividing 10 by 2")
    print(10 / 2)
except ZeroDivisionError:
    print("Error happened!")
finally:
    print("Finally block executed (Case1).")

print("-"*30)

# Case 2: error comes
try:
    print("Case 2: dividing 10 by 0")
    print(10 / 0)
except ZeroDivisionError:
    print("Error happened!")
finally:
    print("Finally block executed (Case 2).")
```

### Q19
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
def check_age(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    return "Age is valid. Access granted."

# Valid age
try:
    print(check_age(20))
except ValueError as e:
    print("Error:",e)

# Invalid age
try:
    print(check_age(15))
except ValueError as e:
    print("Error:",e)
```
